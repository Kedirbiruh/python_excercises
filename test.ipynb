```python
import re
import phonenumbers

def telefonnummer_formatieren(num):
    # normalize input
    if num is None:
        return None
    s = str(num).strip()
    if not s or s.lower() in {"nan", "n/a", "na"}:
        return None

    # remove common separators but keep leading + and leading 00 for country codes
    # first strip spaces, dots, slashes, pipes, hyphens and parentheses
    cleaned = re.sub(r'[.\s\-/|()]+', '', s)

    # handle +49(0)... or 0049(0)... -> remove the extra 0 after country code
    if cleaned.startswith('+490'):
        cleaned = '+49' + cleaned[4:]
    if cleaned.startswith('00490'):
        cleaned = '0049' + cleaned[5:]

    # convert 00... to +...
    if cleaned.startswith('00'):
        cleaned = '+' + cleaned[2:]

    try:
        # prefer explicit international string if present
        if cleaned.startswith('+'):
            parsed = phonenumbers.parse(cleaned, None)
        else:
            parsed = phonenumbers.parse(cleaned, "DE")
        if phonenumbers.is_valid_number(parsed):
            return phonenumbers.format_number(parsed, phonenumbers.PhoneNumberFormat.INTERNATIONAL)
        else:
            return None
    except Exception:
        return None
```

```python
df["Telefonnummer"].apply(telefonnummer_formatieren)
```

```python
df["Telefonnummer"] = df["Telefonnummer"].apply(telefonnummer_formatieren)
df
```
